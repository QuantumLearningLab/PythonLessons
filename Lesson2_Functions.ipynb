Let's start with Functions. While here are some technical terms to describe functions, they can be though of as:


1.   They name a piece of code
2.   They can take in parameters / arguments
3.   They let you make your own "mini scripts" or "tiny commands"

Let's get started by defining a sayHello function. Notice the () after sayHello (which we will use in a little bit to send the function parameters) and the :

```python
def sayHello():
  print("Hello Quantum")
```

Now let's call or use the function

```python
sayHello()
```

Now, let's pass in a name to the function

```python
def sayingHelloTo(name):
  print("Hello",name)
```

And, let's call this new function

```python
person = "Brian"
sayingHelloTo(person)
```

Next we will use the "return" command to send values back from the function. For instance, if we want to calculate the slope of a line between two points:

```python
def getSlope(x1,y1,x2,y2):
  rise = y2-y1
  run = x2-x1
  slope = rise/run
  return slope
```

Now let's call this new function

```python
(x1,y1) = (3,4)
(x2,y2) = (6,12)
theSlope = getSlope(x1,y1,x2,y2)
print(f"Given the points ({x1},{y1}) and ({x2},{y2}) the slope between them is {theSlope}")
```

We can also return one than one value

```python
def doMaths(x,y):
  sum = x+y
  difference = x-y
  return x,y
```

Calling this new function

```python
x = 13
y = 42
s,d = doMaths(x,y)
print(f"Given {x} and {y} the sum is {s} and the difference is {d}")
```
